## Summary of Missing Value Treatment

The missing values in the Finlora Customer Transaction Dataset will be handled according to the type and purpose of each variable:

- **`amount_usd` (305 missing values):** Missing values will be reconstructed using `amount_src` and an appropriate source-to-USD exchange rate after verifying the correct currency conversion relationship.

- **`fee` (295 missing values):** Missing values will be filled using the **median fee for each transaction channel**. Any remaining missing values will be filled using the overall median fee.

- **`ip_country` (301 missing values):** Missing values will be filled using the corresponding `home_country` as a fallback.

- **`kyc_tier` (300 missing values):** Missing values will be filled using the **mode (most frequent value)** of the variable.

- **`device_trust_score` (295 missing values):** Missing values will be filled using the **median `device_trust_score` grouped by `kyc_tier` and `new_device`**. Any remaining missing values will be filled using the overall median.

- **`timestamp` (61 missing values) and `amount_src` (4 missing values):** Records containing missing values in these variables will be removed using `dropna()` because they represent a relatively small proportion of the dataset.

- **`ip_address` (305 missing values):** Conventional imputation will not be applied because IP addresses are identifiers. The variable will be assessed for possible exclusion from the modelling dataset rather than deleting otherwise useful transaction records.

### Conclusion

Overall, missing values will be handled using a combination of **data reconstruction, group-based median imputation, mode imputation, row removal, and possible variable exclusion**, depending on the characteristics and relevance of each variable.

```python
# Calculating the exchange ratee per currency
exchange_rates = (
    customer_transaction_data[
        customer_transaction_data['amount_usd'].notna()
    ]
    .groupby('source_currency')
    .apply(
        lambda x: (x['amount_usd'] / x['amount_src']).mean()
    )
    .to_dict()
)

exchange_rates
```

### Source Currency to USD Conversion Rates

Exchange rates were estimated from transactions with available `amount_usd` values by calculating the ratio of `amount_usd` to `amount_src` and taking the average for each `source_currency`.

The estimated conversion factors were:

| Source Currency | USD Conversion Factor |
|---|---:|
| CAD | 0.7216 |
| GBP | 1.2234 |
| USD | 0.9839 |

These dataset-derived conversion factors will be used for the reconstruction of missing `amount_usd` values.

```python
# Filling in the missing values in the amount_usd column using the calculated exchange rates
customer_transaction_data['amount_usd'] =  customer_transaction_data.apply(
    lambda row:row['amount_usd'] if pd.notna(row['amount_usd']) else row['amount_src'] * exchange_rates[row['source_currency']], axis=1    
)
```

### Imputation of Missing `amount_usd` Values

The `amount_usd` variable contained **305 missing values**. These values were imputed using the transaction's `amount_src` and the dataset-derived USD conversion factor corresponding to each `source_currency`.

Existing `amount_usd` values were retained, while only missing observations were calculated using:

**amount_usd = amount_src × source currency conversion factor**

After imputation, the `amount_usd` variable was reassessed to confirm that the missing values had been successfully treated.

```python
# Filling in the missing values of fee variable
if 'fee' in customer_transaction_data.columns:
    if'channel' in customer_transaction_data.columns:
        customer_transaction_data['fee'] = customer_transaction_data.groupby('channel')['fee'].transform(lambda y: y.fillna(y.median()))
    customer_transaction_data['fee'] = customer_transaction_data['fee'].fillna(customer_transaction_data['fee'].median())
```

```python
# filling in the missing values of device_trust_score variable
if 'device_trust_score' in customer_transaction_data.columns:
    if {'new_device','kyc_tier'}.issubset(customer_transaction_data.columns):
        customer_transaction_data['device_trust_score'] = customer_transaction_data.groupby(['new_device','kyc_tier'])['device_trust_score'].transform(lambda x: x.fillna(x.median()))  
        customer_transaction_data['device_trust_score'] = customer_transaction_data['device_trust_score'].fillna(customer_transaction_data['device_trust_score'].median())
```

```python
# Check the result with:

customer_transaction_data['device_trust_score'].isnull().sum()
```

### Imputation of Missing `device_trust_score` Values

The `device_trust_score` variable contained **295 missing values**. Missing values were first imputed using the **median device trust score within groups defined by `new_device` and `kyc_tier`**. This approach preserves differences in device trust scores across customer verification levels and device status.

Any missing values that could not be filled using the group-specific median were subsequently replaced with the **overall median `device_trust_score`**.

After imputation, the variable was reassessed to confirm that no missing values remained.

```python
# Filling in the missing values of ip_country variable
if {'ip_country', 'home_country'}.issubset(customer_transaction_data.columns):   #checks that both columns exist in the dataset before performing the imputation
    customer_transaction_data['ip_country'] = customer_transaction_data['ip_country'].fillna(customer_transaction_data['home_country'])  # replaces a missing ip_country with the home_country value from the same row.
```
